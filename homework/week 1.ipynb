```python
'''Q6. Median value of horsepower'''
# (i) Find the median value of the horsepower column in the dataset.
df['horsepower'].describe().round(2)
median1 = df['horsepower'].median()
# median=254.00

# (ii) Next, calculate the most frequent value of the same horsepower column.
mode = df['horsepower'].mode()
# mode=252.0

# (iii)Use the fillna method to fill the missing values in the horsepower column with the most frequent value from the previous step.
df['horsepower'] = df['horsepower'].fillna(mode)

# (iv) Now, calculate the median value of horsepower once again.
median2 = df['horsepower'].median()
# median2 = 254.00

# median1 == median2
```

```python
'''Q7. Sum of weights'''
# (i) Select all the cars from Asia
asian_cars = df[df['origin'] == 'Asia']
# df.head()

# (ii) Select only columns vehicle_weight and model_year
array = asian_cars[['vehicle_weight', 'model_year']]

# (iii) Select the first 7 values
x = array[:7]

# (iv) Get the underlying NumPy array. Let's call it X.
x = np.array(x)

# (v) Compute matrix-matrix multiplication between the transpose of X and X. To get the transpose, use X.T. Let's call the result XTX.
Tx = x.T
xTx = Tx.dot(x)
'''x = array([[4240, 1996],
       [4310, 1992],
       [4230, 1997],
       [4420, 1978],
       [3950, 2007],
       [4750, 2021],
       [4450, 2019]])'''
'''Tx = array([[4240, 4310, 4230, 4420, 3950, 4750, 4450],
       [1996, 1992, 1997, 1978, 2007, 2021, 2019]])'''

# (vi) Invert XTX.
xTx_inverse = np.linalg.inv(xTx)
xTx_inverse
# (vii) Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200].
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# (viii) Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w.
w = xTx_inverse.dot(x.T).dot(y)
w

# (ix) What's the sum of all the elements of the result?
w.sum().round(3)
# o.369
```
